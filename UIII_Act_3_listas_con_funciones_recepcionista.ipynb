#### lista con funciones recepcionista
leerrecepcionista(), listarreepcionista(), buscarrecepcionista() , modificarrecepcionista() y borrarrecepcionista(),

```python
# Lista global para almacenar los datos de los recepcionistas [1]
recepcionistas = []

def leerrecepcionista():
    """Solicita datos y los agrega a la lista como un diccionario [2, 3]."""
    nombre = input("Ingrese el nombre del recepcionista: ")
    id_recep = input("Ingrese el ID único: ")
    puesto = input("Ingrese el turno (Mañana/Tarde/Noche): ")

    # Se crea un diccionario y se añade a la lista con append() [2, 4]
    nuevo = {"id": id_recep, "nombre": nombre, "puesto": puesto}
    recepcionistas.append(nuevo)
    print("Recepcionista registrado con éxito.")

def listarrecepcionista():
    """Recorre la lista e imprime cada registro [5]."""
    if not recepcionistas:
        print("No hay recepcionistas registrados.")
        return
    print("\n--- Lista de Recepcionistas ---")
    for r in recepcionistas:
        print(f"ID: {r['id']} | Nombre: {r['nombre']} | Turno: {r['puesto']}")

def buscarecepcionista():
    """Busca un recepcionista por su ID [6, 7]."""
    buscado = input("Ingrese el ID del recepcionista a buscar: ")
    encontrado = False
    for r in recepcionistas:
        if r['id'] == buscado:
            print(f"Encontrado: {r['nombre']} - Turno: {r['puesto']}")
            encontrado = True
            break
    if not encontrado:
        print("Recepcionista no encontrado.")

def modificarrecepcionista():
    """Localiza un registro por ID y permite cambiar sus valores [3, 4]."""
    id_mod = input("Ingrese el ID del recepcionista a modificar: ")
    for r in recepcionistas:
        if r['id'] == id_mod:
            r['nombre'] = input("Nuevo nombre: ")
            r['puesto'] = input("Nuevo turno: ")
            print("Datos actualizados correctamente.")
            return
    print("ID no encontrado.")

def borrarrecepcionista():
    """Elimina un recepcionista de la lista por su ID [8, 9]."""
    id_del = input("Ingrese el ID del recepcionista a borrar: ")
    for i in range(len(recepcionistas)):
        if recepcionistas[i]['id'] == id_del:
            # Se elimina el elemento por índice [8]
            eliminado = recepcionistas.pop(i)
            print(f"Recepcionista {eliminado['nombre']} eliminado.")
            return
    print("No se encontró el ID para eliminar.")

# Bucle principal de control (Menú Interactivo) [10, 11]
while True:
    print("\n--- SISTEMA DE GESTIÓN DE RECEPCIONISTAS ---")
    print("1. Leer (Registrar)")
    print("2. Listar")
    print("3. Buscar")
    print("4. Modificar")
    print("5. Borrar")
    print("6. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        leerrecepcionista()
    elif opcion == "2":
        listarrecepcionista()
    elif opcion == "3":
        buscarecepcionista()
    elif opcion == "4":
        modificarrecepcionista()
    elif opcion == "5":
        borrarrecepcionista()
    elif opcion == "6":
        print("Saliendo del sistema...")
        break  # Rompe el bucle infinito para finalizar el programa [10, 11]
    else:
        print("programa terminado por Ambar Fierro NL = 17")
```
